# tile-math demo

Python API on five European capitals (`examples/cities.geojson`, `[lon, lat]`).

```python
import importlib.util
import subprocess
import sys

if importlib.util.find_spec('tile_math') is None:  # fresh runtime, e.g. Google Colab
    subprocess.check_call([sys.executable, '-m', 'pip', 'install', '-q', 'git+https://github.com/AmmarYasser455/tile-math'])

import json
import os
import urllib.request
from tile_math import lonlat_to_tile, tile_to_quadkey, quadkey_to_tile, tile_bounds, bbox_to_tiles

path = 'examples/cities.geojson'
if os.path.exists(path):
    cities = json.load(open(path, encoding='utf-8'))
else:
    url = 'https://raw.githubusercontent.com/AmmarYasser455/tile-math/main/examples/cities.geojson'
    cities = json.load(urllib.request.urlopen(url))
len(cities['features'])
```

```python
for f in cities['features']:
    lon, lat = f['geometry']['coordinates']
    x, y, z = lonlat_to_tile(lon, lat, 10)
    print(f['properties']['name'], f'{z}/{x}/{y}', tile_to_quadkey(x, y, z))
```

```python
london = next(f for f in cities['features'] if f['properties']['name'] == 'London')
assert lonlat_to_tile(*london['geometry']['coordinates'], 10) == (511, 340, 10)
assert quadkey_to_tile(tile_to_quadkey(511, 340, 10)) == (511, 340, 10)
print(tile_bounds(511, 340, 10))
```

```python
coords = [f['geometry']['coordinates'] for f in cities['features']]
west, east = min(c[0] for c in coords), max(c[0] for c in coords)
south, north = min(c[1] for c in coords), max(c[1] for c in coords)
tiles = bbox_to_tiles(west, south, east, north, 6)
print(len(tiles), tiles[:3])
```
